**Olist** is the connective layer between thousands of small Brazilian merchants and the country's major online marketplaces. Between September 2016 and August 2018, the platform processed **99,441 orders** worth roughly **R$13.6M** in product sales (R$16.0M including freight and before cancellations), growing from a standing start to a steady ~6,500–7,500 orders/month by mid-2018.

The data tells a clear story: **Olist's biggest lever isn't demand — it's fulfillment.** Growth is healthy and broad-based across 27 states and 70+ product categories, but three structural frictions cap the ceiling:

1. **Delivery reliability drives satisfaction almost by itself.** Late orders average a review score of **2.27/5** versus **4.29/5** for on-time orders — a 2-point swing on the single variable Olist can most directly influence through better logistics planning.
2. **Logistics cost scales with geography, not efficiency.** **63.8%** of shipments cross state lines, and cross-state freight costs **~76% more** on average (R$23.69 vs R$13.46) than same-state shipments — pointing to a fulfillment network that isn't yet optimized for seller-to-buyer distance.
3. **Revenue is concentrated in a thin slice of sellers.** The top 10% of sellers generate **~67%** of total product revenue. A platform-level marketplace, by design, is exposed to the performance of a small number of power sellers.

None of these are demand problems — they are **data and decision-system problems**, which is exactly where HVIA's capabilities (forecasting, logistics intelligence, seller analytics, and AI-assisted operations) apply directly. Section 13 turns these three findings into a concrete solution proposal, and Section 14 turns the proposal into an outreach message ready to send.

### 2.2 What Olist Actually Does

Olist is **not a single online store** — it is an integration and infrastructure layer. A small Brazilian retailer that wants to sell on Mercado Livre, Amazon, Shopee, Magalu, and dozens of other marketplaces would otherwise need a separate account, catalog, and logistics setup for each one. Olist collapses that into a single onboarding, giving sellers:

- **Marketplace Hub** — one catalog, listed across 170+ integrated marketplaces
- **Olist Shops (Ecommerce)** — a mobile-first, social-native storefront of the seller's own
- **ERP** — inventory, order, and business management tools built for small operators
- **Olist Pax (Logistics & Shipping)** — a shared fulfillment and shipping network, from warehousing to last-mile delivery
- **Payments & Account / Olist Pay** — payment processing and business banking
- **Business Credit / Olist Credit** — working-capital lending, underwritten using the seller's own sales data
- **AI (IA)** — native AI tooling aimed at helping sellers list, price, and sell more effectively

### 2.3 Who Their Customers Are

Olist's direct customer is the **seller**, not the end shopper — typically a small or medium Brazilian business that lacks the resources to manage multi-channel e-commerce, logistics, and financing independently. The end consumer experience (what this dataset captures) is a *downstream product* of how well Olist serves that seller.

### 2.4 Where the Business Model Creates Value — and Risk

Olist earns from marketplace commissions, subscription/software fees (ERP), logistics margins, payment processing, and credit interest — a genuinely diversified model. But it also means Olist is operationally responsible for a chain it doesn't fully control end-to-end: a purchase made on a third-party marketplace is fulfilled by an independent small seller, shipped through Olist's logistics network, across a country with continental-scale geography. **Every link in that chain is a place where friction shows up in the data we're about to explore.**

### 3.1 Entity Relationship Map

```
customers ──< orders >── order_items >── products ── category_translation
                │              │
                │              └──< sellers
                │
                ├──< order_payments
                └──< order_reviews

geolocation  (joins to customers / sellers via zip_code_prefix)
```

**Grain of each table:**

| Table | Grain (1 row =) | Key columns |
|---|---|---|
| `orders` | one customer order | `order_id`, `customer_id`, `order_status`, purchase/approval/delivery timestamps |
| `order_items` | one product line within an order | `order_id`, `order_item_id`, `product_id`, `seller_id`, `price`, `freight_value` |
| `order_payments` | one payment transaction on an order | `order_id`, `payment_type`, `payment_installments`, `payment_value` |
| `order_reviews` | one customer review of an order | `order_id`, `review_score`, `review_comment_message` |
| `products` | one product listing | `product_id`, `product_category_name`, dimensions/weight |
| `customers` | one customer-order location record | `customer_id`, `customer_unique_id`, `customer_city`, `customer_state` |
| `sellers` | one seller | `seller_id`, `seller_city`, `seller_state` |
| `geolocation` | one zip-code-prefix coordinate | `geolocation_zip_code_prefix`, lat/lng, city, state |
| `product_category_name_translation` | one category name pair | Portuguese → English category name |

An order can contain items from **multiple sellers**, and `customer_id` is order-specific (a repeat shopper gets a new `customer_id` per order, but keeps the same `customer_unique_id`) — both are important join keys to get right, and both are handled explicitly below.


````python

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.ticker as mticker
import matplotlib.patches as mpatches
import seaborn as sns
import warnings
warnings.filterwarnings('ignore')

# ---------------------------------------------------------------
# HVIA brand palette & global chart style
# ---------------------------------------------------------------
NAVY, BLUE, LBLUE, SKY = "#0B1F3A", "#1E6FE0", "#5FA8F5", "#9FD3FF"
TEAL, AMBER, GREY, RED = "#0FA3B1", "#F2A541", "#6B7280", "#E4572E"
PALETTE = [BLUE, LBLUE, NAVY, TEAL, AMBER, SKY]

plt.rcParams.update({
    'font.family': 'DejaVu Sans', 'axes.edgecolor': '#D8E1EC', 'axes.labelcolor': NAVY,
    'text.color': NAVY, 'xtick.color': GREY, 'ytick.color': GREY, 'axes.titleweight': 'bold',
    'figure.facecolor': 'white', 'axes.facecolor': 'white', 'axes.grid': True,
    'grid.color': '#E8EEF5', 'grid.linewidth': 0.8, 'axes.axisbelow': True, 'font.size': 10.5,
})

FIG_DIR = 'figures'
import os
os.makedirs(FIG_DIR, exist_ok=True)

def style_ax(ax, title, xlabel='', ylabel=''):
    ax.set_title(title, fontsize=13.5, color=NAVY, pad=14, fontweight='bold')
    ax.set_xlabel(xlabel, fontsize=10.5)
    ax.set_ylabel(ylabel, fontsize=10.5)
    for s in ['top', 'right']:
        ax.spines[s].set_visible(False)
    return ax

def save(fig, name):
    fig.tight_layout()
    fig.savefig(f'{FIG_DIR}/{name}.png', dpi=300, bbox_inches='tight', facecolor='white')
    plt.show()
    plt.close(fig)

print("Environment ready — HVIA chart style loaded.")
````

````python

# ---------------------------------------------------------------
# Load all 9 tables
# ---------------------------------------------------------------
orders    = pd.read_csv('data/olist_orders_dataset.csv', parse_dates=[
    'order_purchase_timestamp', 'order_approved_at', 'order_delivered_carrier_date',
    'order_delivered_customer_date', 'order_estimated_delivery_date'])
items     = pd.read_csv('data/olist_order_items_dataset.csv', parse_dates=['shipping_limit_date'])
payments  = pd.read_csv('data/olist_order_payments_dataset.csv')
reviews   = pd.read_csv('data/olist_order_reviews_dataset.csv')
products  = pd.read_csv('data/olist_products_dataset.csv')
customers = pd.read_csv('data/olist_customers_dataset.csv')
sellers   = pd.read_csv('data/olist_sellers_dataset.csv')
cat_trans = pd.read_csv('data/product_category_name_translation.csv')
geo       = pd.read_csv('data/olist_geolocation_dataset.csv')

tables = {
    'orders': orders, 'order_items': items, 'order_payments': payments,
    'order_reviews': reviews, 'products': products, 'customers': customers,
    'sellers': sellers, 'category_translation': cat_trans, 'geolocation': geo,
}
summary = pd.DataFrame([(name, df.shape[0], df.shape[1]) for name, df in tables.items()],
                        columns=['table', 'rows', 'columns']).sort_values('rows', ascending=False)
summary
````

Three deliberate choices shape every result in this notebook:

1. **Scope to `delivered` orders for delivery/review analysis.** Undelivered orders (shipped, canceled, unavailable, processing) have no delivery timestamp and would bias delivery-time statistics if included.
2. **Drop the final, incomplete partial month.** Order volume collapses to single digits in September–October 2018 because the export simply stops mid-month, not because demand fell off a cliff. Trend charts are scoped to complete months (through August 2018) so growth isn't misread as decline.
3. **Join on `product_category_name_english`** wherever a category is shown, so every chart is legible without a Portuguese-English lookup step.


````python

# ---------------------------------------------------------------
# Master analytical table: one row per order item, enriched
# ---------------------------------------------------------------
df = (items
      .merge(orders, on='order_id', how='left')
      .merge(customers, on='customer_id', how='left')
      .merge(sellers, on='seller_id', how='left')
      .merge(products, on='product_id', how='left')
      .merge(cat_trans, on='product_category_name', how='left'))

# Delivery + lateness features (only meaningful for delivered orders)
df['delivery_days'] = (df['order_delivered_customer_date'] - df['order_purchase_timestamp']).dt.days
df['delay_vs_estimate_days'] = (df['order_delivered_customer_date'] - df['order_estimated_delivery_date']).dt.days
df['is_late'] = df['delay_vs_estimate_days'] > 0
df['is_cross_state'] = df['customer_state'] != df['seller_state']
df['order_month'] = df['order_purchase_timestamp'].dt.to_period('M').dt.to_timestamp()

# Complete-months only (excludes the truncated Sep/Oct 2018 tail)
df_complete = df[df['order_purchase_timestamp'] < '2018-09-01'].copy()
delivered = df[df['order_status'] == 'delivered'].copy()

print(f"Master table: {df.shape[0]:,} order-item rows across {df['order_id'].nunique():,} orders")
print(f"Delivered subset: {delivered.shape[0]:,} rows across {delivered['order_id'].nunique():,} orders")
print(f"Date range (complete months): {df_complete.order_purchase_timestamp.min().date()} to {df_complete.order_purchase_timestamp.max().date()}")
````


<a id="growth"></a>
<div style="background:#0B1F3A; color:white; padding:16px 22px; border-radius:8px; font-family:'Segoe UI',Arial,sans-serif; margin-top:6px;">
  <div style="font-size:12px; letter-spacing:3px; color:#5FA8F5; font-weight:600;">SECTION 5</div>
  <div style="font-size:24px; font-weight:800; margin-top:4px;">Growth & Revenue Performance</div>
  <div style="font-size:14px; color:#CFE4FF; margin-top:4px;">GMV trend, order volume, and average order value</div>
</div>


````python

monthly = (df_complete.groupby('order_month')
           .agg(gmv=('price', 'sum'), orders=('order_id', 'nunique'), items=('order_item_id', 'count'))
           .reset_index())
monthly['aov'] = monthly['gmv'] / monthly['orders']

fig, ax1 = plt.subplots(figsize=(12, 5.2))
ax1.bar(monthly.order_month, monthly.gmv/1000, color=BLUE, width=20, zorder=3, label='GMV (R$ thousand)')
ax1.yaxis.set_major_formatter(mticker.FuncFormatter(lambda x, p: f'R${x:,.0f}k'))
style_ax(ax1, 'Monthly Gross Merchandise Value & Order Volume', 'Month', 'GMV (R$ thousand)')
ax2 = ax1.twinx()
ax2.plot(monthly.order_month, monthly.orders, color=NAVY, marker='o', linewidth=2.2, markersize=4.5, label='Orders')
ax2.set_ylabel('Number of Orders', fontsize=10.5, color=NAVY)
ax2.grid(False)
fig.autofmt_xdate()
l1, lb1 = ax1.get_legend_handles_labels(); l2, lb2 = ax2.get_legend_handles_labels()
ax1.legend(l1+l2, lb1+lb2, loc='upper left', frameon=False)
save(fig, '01_monthly_gmv_orders')

total_gmv = df_complete.price.sum()
total_freight = df_complete.freight_value.sum()
total_orders = df_complete.order_id.nunique()
aov = total_gmv / total_orders
print(f"Total product GMV (complete months): R$ {total_gmv:,.2f}")
print(f"Total freight collected: R$ {total_freight:,.2f}")
print(f"Total distinct orders: {total_orders:,}")
print(f"Average Order Value (product only): R$ {aov:,.2f}")
print(f"Peak month: {monthly.loc[monthly.gmv.idxmax(),'order_month'].strftime('%B %Y')} with R$ {monthly.gmv.max():,.0f}")
````


**Reading the chart:** GMV and order count move almost in lockstep — growth is driven by *more orders*, not materially higher spend per order, which stays fairly flat around the low-R$100s. The **November 2017 spike** aligns with Black Friday, showing Olist's sellers are already capitalizing on seasonal demand peaks — a signal that peak-period logistics capacity (Section 6) matters disproportionately.

````python

fig, ax = plt.subplots(figsize=(11, 5))
sns.histplot(delivered.delivery_days.dropna(), bins=45, color=BLUE, ax=ax, edgecolor='white')
med = delivered.delivery_days.median()
ax.axvline(med, color=NAVY, linestyle='--', linewidth=1.8, label=f'Median: {med:.0f} days')
style_ax(ax, 'Distribution of Delivery Time (Purchase → Customer Delivery)', 'Delivery Time (days)', 'Number of Orders')
ax.legend(frameon=False)
save(fig, '02_delivery_time_distribution')

pct_late = delivered.is_late.mean() * 100
avg_days = delivered.delivery_days.mean()
avg_delay_when_late = delivered.loc[delivered.is_late, 'delay_vs_estimate_days'].mean()
print(f"Average delivery time: {avg_days:.1f} days (median {med:.0f})")
print(f"Orders delivered AFTER the estimated date: {pct_late:.1f}%")
print(f"Average delay when late: {avg_delay_when_late:.1f} days past the promised date")
````

````python

# Cross-state vs same-state freight economics
fr = delivered.dropna(subset=['seller_state', 'customer_state']).copy()
freight_cmp = fr.groupby('is_cross_state').agg(
    avg_freight=('freight_value', 'mean'), avg_price=('price', 'mean'), orders=('order_id', 'nunique')
).rename(index={False: 'Same State', True: 'Cross State'})
pct_cross_state = fr.is_cross_state.mean() * 100

rev_del = delivered.dropna(subset=['order_id']).drop_duplicates('order_id')[['order_id', 'is_late']].merge(
    reviews[['order_id', 'review_score']], on='order_id', how='inner')
score_by_late = rev_del.groupby('is_late').review_score.mean()

fig, axes = plt.subplots(1, 2, figsize=(12, 5))
sns.barplot(x=['On-Time', 'Late'], y=[score_by_late.get(False, 0), score_by_late.get(True, 0)],
            palette=[BLUE, RED], ax=axes[0])
style_ax(axes[0], 'Avg. Review Score: On-Time vs Late Delivery', '', 'Average Review Score (1–5)')
axes[0].set_ylim(0, 5)
for i, v in enumerate([score_by_late.get(False, 0), score_by_late.get(True, 0)]):
    axes[0].text(i, v + 0.12, f'{v:.2f}', ha='center', fontweight='bold', color=NAVY)

sns.barplot(x=freight_cmp.index, y=freight_cmp.avg_freight, palette=[BLUE, AMBER], ax=axes[1])
style_ax(axes[1], f'Avg. Freight Cost: Same-State vs Cross-State\n({pct_cross_state:.0f}% of shipments cross state lines)', '', 'Average Freight (R$)')
for i, v in enumerate(freight_cmp.avg_freight):
    axes[1].text(i, v + 0.4, f'R${v:.2f}', ha='center', fontweight='bold', color=NAVY)
save(fig, '03_delivery_and_freight_impact')

print(f"Cross-state shipments: {pct_cross_state:.1f}% of all deliveries")
print(freight_cmp)
````


**Reading the charts:** the review-score gap (left) is the single strongest relationship in this entire dataset — a **2-point swing** on a 5-point scale attributable to one operational variable. The freight comparison (right) shows *why* delivery slips in the first place: nearly two-thirds of all shipments already cross state lines in a country the size of a continent, and each one costs ~76% more to ship. Late delivery isn't random noise — it's the predictable outcome of a seller network that isn't matched to buyer geography.

````python

score_counts = reviews.review_score.value_counts().sort_index()
fig, ax = plt.subplots(figsize=(9, 5))
bars = ax.bar(score_counts.index.astype(str), score_counts.values,
              color=[RED, AMBER, GREY, LBLUE, BLUE])
style_ax(ax, 'Review Score Distribution (1 = worst, 5 = best)', 'Review Score', 'Number of Reviews')
for b in bars:
    h = b.get_height()
    ax.text(b.get_x()+b.get_width()/2, h + 800, f'{h:,}\n({h/score_counts.sum()*100:.0f}%)',
            ha='center', fontsize=9.5, color=NAVY)
save(fig, '04_review_score_distribution')

avg_score = reviews.review_score.mean()
pct_5star = (reviews.review_score == 5).mean() * 100
pct_1star = (reviews.review_score == 1).mean() * 100
pct_with_comment = reviews.review_comment_message.notna().mean() * 100
print(f"Average review score: {avg_score:.2f} / 5")
print(f"5-star share: {pct_5star:.1f}%   |   1-star share: {pct_1star:.1f}%")
print(f"Reviews with a written comment: {pct_with_comment:.1f}%")
````


**Reading the chart:** the distribution is bimodal — a large 5-star majority alongside a disproportionately large 1-star tail (bigger than the 2-star bucket and comparable to the 3-star bucket). That shape is consistent with a satisfaction driver that is binary in practice (delivered acceptably vs. failed badly) rather than a smooth quality gradient — which lines up exactly with the on-time/late split in Section 6. **Roughly 4 in 10 reviews include a written comment** — an underused free-text signal for automated issue-tagging (see Section 13).

No single seller dominates (the largest is under 2% of revenue), so the risk is *distributional*, not a single-point-of-failure — which is actually good news for a churn-prevention strategy aimed at the broad "long tail."

  <div style="font-size:14px; color:#CFE4FF; margin-top:4px;">What is really happening at Olist, told as one narrative</div>
</div>



Pulling every section together, one coherent story emerges:

### The growth is real — and it's outrunning the fulfillment network built to support it

Olist scaled from a standing start to ~7,000 orders/month in under two years, spread across 27 states and dozens of categories with no single point of concentration risk on the demand side. That is a genuinely healthy growth curve for a marketplace-enablement business.

### But the seller base is geographically lopsided relative to demand

Nearly 60% of sellers are based in São Paulo state, while only 42% of customers are. That gap is structural — most of the country is, by definition, being served from far away. It shows up directly as a **63.8% cross-state shipment rate**, and cross-state shipping costs **~76% more** on average than local fulfillment.

### That geography gap is the direct cause of late deliveries — and late deliveries are the direct cause of bad reviews

Only 6.6% of delivered orders arrive after the promised date, but for that minority, the damage to customer sentiment is severe: average review score falls from **4.29 to 2.27** — nearly a 2-point collapse on a 5-point scale. Because Olist's own AI (per its product positioning) is meant to drive more sales, and marketplaces weight seller reputation heavily in ranking algorithms, **a small percentage of badly-managed deliveries has an outsized effect on long-term seller visibility and sales**, not just on the one order.

### The platform's revenue depends more on its top sellers than growth metrics alone would suggest

The top 10% of sellers generate two-thirds of all revenue. That's not an alarming single-point-of-failure (the single biggest seller is under 2% of revenue), but it does mean the **health of the "long tail"** — the other 90% of sellers, presumably the newer, smaller, and less sophisticated ones — is where the platform's future growth ceiling actually lives. Helping that long tail sell more, price better, and ship faster is a bigger lever than acquiring more top sellers.

### None of this requires guessing — it requires better decision infrastructure

Every friction identified above is measurable in the data Olist already collects. What's missing isn't information; it's the systematic, real-time translation of that information into seller-facing decisions (who to route an order to, which seller needs help, when to warn a customer about a realistic delivery window). **That gap is precisely what HVIA is positioned to close.**

**Category:** Logistics Intelligence + Forecasting & Predictive Solutions

**The offer:** a predictive model, trained on Olist's own historical order/seller/route data, that scores **every order at checkout** for delivery-risk (probability of arriving late, given seller location, buyer location, carrier, season, and category). ###  Proposal 3 — AI-Assisted Review & Complaint Triage
**Category:** AI-Assisted Operations + Automation & Data Pipelines

**The evidence:** ~41% of reviews include a free-text comment that is currently unused for anything but display.

**The offer:** an NLP pipeline that classifies review comments by root cause (late delivery, product quality, wrong item, customer service) in real time, routes the negative ones to the right team automatically, and feeds a recurring "top complaint drivers" report — replacing manual reading of tens of thousands of comments with an always-on triage layer.

````python

from openpyxl import Workbook
from openpyxl.styles import Font, PatternFill, Alignment, Border, Side
from openpyxl.utils import get_column_letter
from openpyxl.chart import BarChart, LineChart, Reference

wb = Workbook()
wb.remove(wb.active)

HEADER_FILL = PatternFill(start_color='0B1F3A', end_color='0B1F3A', fill_type='solid')
HEADER_FONT = Font(color='FFFFFF', bold=True, size=11, name='Calibri')
TITLE_FONT = Font(color='0B1F3A', bold=True, size=14, name='Calibri')
SUB_FONT = Font(color='1E6FE0', bold=True, size=10, italic=True, name='Calibri')
BODY_FONT = Font(name='Calibri', size=10.5, color='1A1A1A')
THIN = Side(style='thin', color='D8E1EC')
BORDER = Border(left=THIN, right=THIN, top=THIN, bottom=THIN)
ALT_FILL = PatternFill(start_color='F4F8FC', end_color='F4F8FC', fill_type='solid')

def write_sheet(wb, sheet_name, title, subtitle, dataframe, number_formats=None):
    ws = wb.create_sheet(sheet_name)
    ws['A1'] = title
    ws['A1'].font = TITLE_FONT
    ws['A2'] = subtitle
    ws['A2'].font = SUB_FONT
    start_row = 4
    for j, col in enumerate(dataframe.columns, start=1):
        c = ws.cell(row=start_row, column=j, value=str(col).replace('_', ' ').title())
        c.font = HEADER_FONT
        c.fill = HEADER_FILL
        c.alignment = Alignment(horizontal='center', vertical='center')
        c.border = BORDER
    for i, (_, row) in enumerate(dataframe.iterrows(), start=1):
        for j, col in enumerate(dataframe.columns, start=1):
            val = row[col]
            c = ws.cell(row=start_row + i, column=j, value=val)
            c.font = BODY_FONT
            c.border = BORDER
            if i % 2 == 0:
                c.fill = ALT_FILL
            if number_formats and col in number_formats:
                c.number_format = number_formats[col]
    for j, col in enumerate(dataframe.columns, start=1):
        max_len = max([len(str(col))] + [len(f'{v:,.2f}') if isinstance(v, float) else len(str(v)) for v in dataframe[col]])
        ws.column_dimensions[get_column_letter(j)].width = min(max(max_len + 4, 12), 40)
    ws.freeze_panes = ws.cell(row=start_row + 1, column=1)
    ws.row_dimensions[1].height = 22
    return ws

# --- Sheet 1: Executive KPIs ---
kpi_data = pd.DataFrame({
    'Metric': ['Total Orders (complete months)', 'Total Product GMV (R$)', 'Total Freight (R$)',
               'Average Order Value (R$)', 'Average Delivery Time (days)', 'Late Delivery Rate (%)',
               'Avg Review Score — On-Time', 'Avg Review Score — Late', 'Cross-State Shipment Rate (%)',
               'Avg Freight — Same State (R$)', 'Avg Freight — Cross State (R$)',
               'Total Active Sellers', 'Top 10% Sellers — Revenue Share (%)', 'Average Review Score (overall)'],
    'Value': [total_orders, round(total_gmv, 2), round(total_freight, 2), round(aov, 2),
              round(avg_days, 1), round(pct_late, 1), round(score_by_late.get(False, 0), 2),
              round(score_by_late.get(True, 0), 2), round(pct_cross_state, 1),
              round(freight_cmp.loc['Same State', 'avg_freight'], 2),
              round(freight_cmp.loc['Cross State', 'avg_freight'], 2),
              total_sellers, round(top10_cum, 1), round(avg_score, 2)]
})
write_sheet(wb, 'Executive_Summary', 'HVIA x Olist — Executive KPI Summary',
            'Task 01 — Olist Business Discovery · All figures from the public Olist dataset', kpi_data)

# --- Sheet 2: Monthly Sales ---
monthly_out = monthly.copy()
monthly_out['order_month'] = monthly_out['order_month'].dt.strftime('%Y-%m')
write_sheet(wb, 'Monthly_Sales', 'Monthly GMV & Order Volume', 'Sep 2016 – Aug 2018 (complete months only)',
            monthly_out, number_formats={'gmv': '#,##0.00', 'aov': '#,##0.00'})

# --- Sheet 3: Delivery Performance ---
delivery_summary = pd.DataFrame({
    'segment': ['Same State', 'Cross State'],
    'avg_freight_r$': freight_cmp['avg_freight'].round(2).values,
    'avg_price_r$': freight_cmp['avg_price'].round(2).values,
    'orders': freight_cmp['orders'].values,
})
write_sheet(wb, 'Delivery_Performance', 'Delivery & Freight Performance',
            f'On-time avg score {score_by_late.get(False,0):.2f} vs late avg score {score_by_late.get(True,0):.2f}',
            delivery_summary)

# --- Sheet 4: Reviews ---
reviews_out = score_counts.reset_index()
reviews_out.columns = ['review_score', 'count']
reviews_out['pct'] = (reviews_out['count'] / reviews_out['count'].sum() * 100).round(1)
write_sheet(wb, 'Reviews_Analysis', 'Review Score Distribution', f'Average score: {avg_score:.2f}/5', reviews_out)

# --- Sheet 5: Category Performance ---
cat_out = cat_perf.reset_index().round(2)
write_sheet(wb, 'Category_Performance', 'Top 10 Categories by Revenue', 'Ranked by total product revenue (R$)',
            cat_out, number_formats={'revenue': '#,##0.00', 'avg_price': '#,##0.00'})

# --- Sheet 6: Geography ---
geo_out = state_perf.reset_index().round(2)
write_sheet(wb, 'Geography_State', 'Top 10 States by Customer Revenue', 'Customer state, ranked by revenue (R$)',
            geo_out, number_formats={'revenue': '#,##0.00'})

# --- Sheet 7: Seller Concentration ---
seller_conc_out = seller_rev.head(50)[['rank', 'seller_id', 'price', 'cum_pct']].round(2)
seller_conc_out.columns = ['rank', 'seller_id', 'revenue_r$', 'cumulative_pct_of_total']
write_sheet(wb, 'Seller_Concentration', 'Top 50 Sellers by Revenue',
            f'Top 10% of sellers = {top10_cum:.1f}% of total platform revenue', seller_conc_out,
            number_formats={'revenue_r$': '#,##0.00'})

# --- Sheet 8: Payments ---
pay_out = pay_type.reset_index()
pay_out.columns = ['payment_type', 'count']
pay_out['pct'] = (pay_out['count'] / pay_out['count'].sum() * 100).round(1)
write_sheet(wb, 'Payments', 'Payment Method Mix', f'Avg. credit card installments: {installments.mean():.1f}', pay_out)

# --- Native Excel chart on Monthly_Sales ---
ws_m = wb['Monthly_Sales']
chart = LineChart()
chart.title = "Monthly GMV Trend"
chart.style = 2
chart.y_axis.title = 'GMV (R$)'
chart.x_axis.title = 'Month'
data_ref = Reference(ws_m, min_col=2, min_row=4, max_row=4+len(monthly_out))
cats_ref = Reference(ws_m, min_col=1, min_row=5, max_row=4+len(monthly_out))
chart.add_data(data_ref, titles_from_data=True)
chart.set_categories(cats_ref)
chart.width, chart.height = 18, 9
ws_m.add_chart(chart, f'{get_column_letter(len(monthly_out.columns)+2)}4')

wb.save('HVIA_Olist_Analysis_Results.xlsx')
print("Excel workbook saved: HVIA_Olist_Analysis_Results.xlsx")
print(f"Sheets: {wb.sheetnames}")
````
